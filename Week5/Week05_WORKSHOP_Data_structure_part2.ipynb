## 1. `List`
List items are **ordered**, **changeable**, and **allow duplicate** values.

- **Ordered** means that the items have a defined order and that order will not change.
If you add new items to a list, the new items will be placed at the end of the list.
(Note: Some list methods will change the order, but in general, the order of the items will not change)
- **Changeable**: We can change, add, and remove items from a list after creation.
- **Allow duplicate** values: lists can have items with the same value.

**Common uses:**

* Keeping an ordered collection of data (e.g., names, numbers).

* When items might change or grow over time.

* Iterating, sorting,

### Common List Operations

````python
# Creating a list

numbers = [1, 3, 4, 5]

# index    0  1  2  3
# neg index     -2 -1
````

````python
# Adding an element

numbers.append(7)
````

````python
# Removing by value

numbers.remove(1)
````

````python
# Removing last value 

numbers.pop()
````

````python
# Del elemnt by index

del numbers[0]
````

````python
# Del all avlaues

numbers.clear()
````

````python
# Update

numbers.insert(1,100)
````

````python
# Print all values

print(numbers)
````

````python
# Accessing one element by index

print(numbers[0])  
````

````python
# Iterate over each elements by value

for i in numbers:
    print(i)
````

````python
# Gen number of elemnts in the list

len(numbers)
````

````python
# Iterate over elemnts by index

for i in range(len(numbers)):
    print(i, numbers[i])
````

````python
# Enumarete elemnts 

for i, j in enumerate(numbers):
    print(i, j)
````

## 2. `Tuple`

Tuple items are **ordered**, **unchangeable (immutable)**, and **allow duplicate** values.

* **Ordered:** Tuples maintain the order of items.
  The order of elements will not change unless a new tuple is created.

* **Immutable:** Once a tuple is created, its items **cannot be changed**, **added**, or **removed**.
  (You must create a new tuple to modify the content.)

* **Allow duplicate values:** Tuples can contain items with the same value.



**Common uses:**

* Storing **constant data** like coordinates, colors, or dates.
* **Returning multiple values** from a function.
* Using as **dictionary keys** (since tuples are immutable).
* Protecting data that shouldn’t be modified.



````python
# Creating a tuple

x = (1, 2, 3)
````

## 3. `Set`

A **set** is a collection of **unique (distinct) items**.
Sets are **unordered** and **mutable**, meaning you can add or remove items — but not change their order or use indexing.

* **Unique values:** A set automatically removes duplicates.
* **Unordered:** The items have no defined order, so you cannot access them by index.
* **Mutable:** You can add or remove items after creation using set methods.
* **Useful for:** Handling large datasets where duplicate values should be ignored or for performing mathematical operations like union, intersection, and difference.

---


###  When to Use Sets

Use a **set** when you need to store **unique items** and don’t care about their order.

**Common uses:**

* Removing duplicates

* Checking membership quickly

* Performing mathematical operations (union, intersection, difference)

````python
# Creating a set
x = {3, 5, 3, 3}
print("Set with unique elements:", x) 
````

````python
# Converting a list to a set to remove duplicates
list1 = [1, 2, 1, 4]
x = set(list1)
print("Set from list:", x)  # Output: {1, 2, 4}
````

````python
# Basic set operations
x = {18, 21, 24, 27}
````

````python
# Adding an item
x.add(30)
print("After add:", x)  # Output: {18, 21, 24, 27, 30}
````

````python
# Removing an item
x.remove(18)
print("After remove:", x)  # Output: {21, 24, 27, 30}
````

````python
# Check membership
print("21 in set:", 21 in x)  # Output: True
````

````python
# Remove last element
x.pop()
print(x)
````

### List of functions on sets:

````python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
````

````python
### Intersection ( `AND` )
````

````python
print(set1 & set2 )
````

````python
print(set1.intersection(set2))
````

````python
### Union ( `OR` )
````

````python
print(set1 | set2)
````

````python
print(set1.union(set2))
````

### Symmetric Difference ( `XOR` )

````python
print( set1 ^ set2)
````

````python
print( set1.symmetric_difference(set2))
````

### Difference (In set1 but not in set2)

````python
print(set1 - set2)
````

````python
print(set2 - set1)
````

````python
print(set1.difference(set2))
````

````python
print(set1)
set1.discard(3)
print(set1)
````

## 4. `Dictionary`

A **dictionary** stores data as **key–value pairs**, allowing you to associate related information.
Each **key** must be unique, and it maps to a specific **value**.

* **Ordered (Python 3.7+):** Items maintain insertion order.
* **Mutable:** You can change, add, or remove key–value pairs.
* **Unique keys:** Keys must be unique; values can be duplicated.
* **Useful for:** Representing structured data (like records, configurations, or JSON-like data).



**Common uses:**

* Storing and accessing **related data** (like student records or configurations).
* **Fast lookups** using a key (faster than searching a list).
* Representing **structured data** like JSON objects or database records.
* Counting or grouping items efficiently.


````python
# Creating a dictionary with a single key-value pair
student = {
    'id': 101,
    'name': 'Alice',
    'age': 20,
    'course': 'Computer Science'
}
student
````

````python
# Creating a dictionary using the dict() function with a list of tuples
student = dict([
    ('id', 101),
    ('name', 'Alice'),
    ('age', 20),
    ('course', 'Computer Science')
])

print(student)
````

````python
# Creating a dictionary using the dict() function with keyword arguments
student = dict(id=101, name='Alice', age=20, course='Computer Science')

print(student)
````

````python

# Update an existing key or add a new one
student['age'] = 21               # Updates existing key
student['grade'] = 'A'            # Adds a new key-value pair

print("Updated student record:", student)
````

````python
# Remove a key-value pair using del
del student['grade']

print("After deletion:", student)
````

````python

# Remove the last key-value pair
student.popitem()
````

````python
# Check if a key exists
print('name' in student)     # True
print('grade' in student)    # False
````

````python
# Check that a key does NOT exist
print('grade' not in student)   # True
print('name' not in student)    # False
````

### Accessing elements in a dictionary

````python
# Returns a view of all keys in the dictionary
print(student.keys())
````

````python
# Returns a view of all values in the dictionary

print(student.values())
````

````python
# Returns all key-value pairs in the dictionary as tuples

print(student.items())
````

````python
# Iterate through key-value pairs in a dictionary

for key, value in student.items():
    print(key, ":", value)
````

````python
# Iterating over key-value pairs in a dictionary

for key, value in student.items():
    print(f"{key}: {value}")
````

````python
# Iterate over only the keys in a dictionary

for key, value in student.items():
    print(key)
````

````python
# Iterate over keys in the dictionary

for k, v in student.items():
    print(k)
````

### Importing Pandas

### Creating Data Structures

## Series

* A **Series** is a **one-dimensional array-like object** that can hold data of any type.

````python
# Creating a Series

data = [10, 20, 30, 40]
index = ['a', 'b', 'c', 'd']

series = pd.Series(data, index=index)
print(series)
````

### DataFrame

* A **DataFrame** is a **two-dimensional, table-like data structure **with labeled axes (rows and columns).

````python
# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data)
print(df)
````
