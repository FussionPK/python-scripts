## `I. File`

## Files


* **Files** allow programs to **store and retrieve data permanently** on a disk.
* A **text file** consists of a series of readable characters.
* A **binary file** (used for images, videos, etc.) consists of a series of bytes.
* The first character in a text file, or byte in a binary file, is at **position 0**.

  * In a file with *n* characters or bytes, the last position is **n − 1**.
* When you open a file in Python using the `open()` function, it returns a **file object** that allows you to interact with the file.

## `Text and CSV Files`

### Text Files

* A **text file** stores data as **readable characters** arranged in lines.
* Each line ends with a **newline character (`\n`)**.
* Common extensions: `.txt`, `.log`, `.cfg`, `.py`.
* Text files use character encodings such as **UTF-8** or **ASCII**.
* They are commonly used for **logs**, **configuration files**, and **plain text data**.


---

### CSV Files

* A **CSV (Comma-Separated Values)** file stores **tabular data** as plain text.
* Each **row** represents a record, and each **column** is separated by a **delimiter** (commonly a comma).
* Variants may use **semicolons (;)** or **tabs (\t)** as separators.
* Common extension: `.csv`.
* Easy to open in **Excel**, **Google Sheets**, or **Python**.

## `II. File Handling Function`

* **`open(file_path, mode)`** → Opens a file and returns a **file object**.
<br>

* **Parameters:**

  * `file_path`: The name or path of the file.
  * `mode`: The mode in which the file is opened (`'r'`, `'w'`, `'a'`, `'rb'`, etc.).

**Example:**
```python
file = open("path/file.txt", "r")   # Opens the file in read mode
```


---

### `file paths`

The **file path** tells Python **where to find or save a file** on your computer.
Different operating systems use different path formats and rules.

| System                   | Example Path                             | Notes                                                                          |
| ------------------------ | ---------------------------------------- | ------------------------------------------------------------------------------ |
| **Windows**              | `"C:\\Users\\alex\\Documents\\data.txt"` | Uses **backslashes (`\`)**. Must be **escaped** (`\\`) or use **raw strings**. |
| **Windows (raw string)** | `r"C:\Users\alex\Documents\data.txt"`    | The `r` prefix tells Python **not to treat backslashes as escape characters**. |
| **Mac**                  | `"/Users/alex/Documents/data.txt"`       | Uses **forward slashes (`/`)** and is **case-sensitive**.                      |
| **Linux**                | `"/home/alex/Documents/data.txt"`        | Uses **forward slashes (`/`)** and is **case-sensitive**.                      |

<i>**Note:**</i>
If your Python file and the data file are in the same directory,
you don’t need to include the full path — just the file name is enough.

---

### `mode` 

The **`mode`** in `open(path, mode)` tells Python **how** you want to use (access) the file.

| Mode   | Meaning           | Description                                                                    |
| ------ | ----------------- | ------------------------------------------------------------------------------ |
| `'r'`  | **Read**          | Opens an existing file for reading only (error if file doesn’t exist).         |
| `'w'`  | **Write**         | Opens for writing — **overwrites** existing content or **creates** a new file. |
| `'a'`  | **Append**        | Opens for writing — adds data to the **end** of the file or creates a new one. |
| `'r+'` | **Read & Write**  | Opens for both reading and writing (file **must exist**).                      |
| `'w+'` | **Write & Read**  | Opens for both, but **clears** the file first or creates a new one.            |
| `'a+'` | **Append & Read** | Opens for both, adds new data to the end, or creates a new file.               |


<i>**Note:**</i>

* Add a `'b'` after any mode (e.g., `'rb'`, `'wb'`) to handle **binary files** such as images, executables, or PDFs.
* Always close the file after use, or use a `with` statement for automatic closing.

### a. Writing a text to a File

- Python’s file handling capabilities are built on the [**`io` module**](https://docs.python.org/3/library/io.html), which provides a range of classes for working with files.

* The **`with` statement** simplifies file handling by automatically managing resources.

  * Opens (acquires) a file and assigns it to a variable.
  * Lets you safely use the file within an indented block.
  * Automatically calls the file object’s **`close()`** method when the block ends — even if an error occurs.

```python
with open("example.txt", "w") as file:
    file.write("Hello, world!")
# File automatically closed here
```


````python
# Example:
# Write account data to a text file
with open("accounts.txt", mode="w") as f:
    f.write("101 Smith 150.75\n")
    f.write("202 Johnson 89.34\n")
    f.write("303 Williams 452.00\n")
    f.write("404 Brown -23.45\n")
    f.write("505 Davis 320.89\n")

print('done ')
````

## III `File` Object — Common Methods

Python’s **file objects** provide several methods for reading from and writing to files.
(Reference: [Python `io` module documentation](https://docs.python.org/3/library/io.html))

| Method         | Description                                                     |
| -------------- | --------------------------------------------------------------- |
| `write()`      | Writes a single string to the file.                             |
| `writelines()` | Writes a list (or iterable) of strings to the file.             |
| `read()`       | Reads the **entire** file content as one string.                |
| `readline()`   | Reads **one line** from the file at a time.                     |
| `readlines()`  | Reads the **entire file** into a list of strings, one per line. |



## **Examples:**

* Using `write()` → adds one string at a time to the file. You must include `\n` manually if you want a new line after each entry.

````python
# Using write() to add data to a file
with open("accounts.txt", "w") as f:
    f.write("101 Smith 150.75\n")
    f.write("202 Johnson 89.34\n")
    f.write("303 Williams 452.00\n")

print("Data written successfully using write().")
````

* Using `writelines()`→ writes all strings in a list or iterable to the file at once. Each line in the list must include a `\n` at the end to appear on a new line.

````python
# Using writelines() to write multiple lines from a list
records = [
    "101 Smith 150.75\n",
    "202 Johnson 89.34\n",
    "303 Williams 452.00\n"
]

with open("accounts.txt", "w") as f:
    f.writelines(records)

print("Data written successfully using writelines().")
````

* Using `read(5)` → Reading Specific Number of Characters

````python
# Read only the first 5 characters from a file
with open("accounts.txt", "r") as f:
    data = f.read(5)
print(data)
````

* Using `read()` → Reads the entire file text as a single <u>string</u>.

````python
with open("accounts.txt", "r") as f:
    data = f.read()
print(data)
````

* Using `readline()` → Reads <u>one line</u> at a time (useful for looping through lines).

````python
with open("accounts.txt", "r") as f:
    line1 = f.readline()
    line2 = f.readline()
print(line1)
print(line2)
````

* Using `readlines()` → Reads all lines into a <u>list</u> of strings.

````python
with open("accounts.txt", "r") as f:
    lines = f.readlines()
print(lines)
````

## IV Handling File Errors

Sometimes, a program tries to open a file that doesn’t exist or isn’t accessible.
In such cases, Python raises a **`FileNotFoundError`** or other exceptions.
To prevent the program from crashing, you can use a **`try–except`** block to handle the error gracefully.


### Explanation

* **`try` block:** Runs the code that may raise an error.
* **`except FileNotFoundError:`** Catches the error and prevents the program from crashing.
* **`open(..., 'w')` inside except:** Creates the file if it doesn’t exist.
* **Program continues safely** — reading or printing data afterward.



````python
# Example

try:
    with open("accounts.txt", "r") as f:
        data = f.read()

except FileNotFoundError:
    print("Error: accounts.txt not found. Creating a new file...")
    # Create the file if it doesn't exist
    with open("accounts.txt", "w") as f:
        f.write("101 Smith 150.75\n")
        f.write("202 Johnson 89.34\n")
        f.write("303 Williams 452.00\n")

    # Reopen and read after creating
    with open("accounts.txt", "r") as f:
        data = f.read()

# Display file contents
print(f"\n{'Account':<10}{'Name':<10}{'Balance':>10}")
print(data)
````
